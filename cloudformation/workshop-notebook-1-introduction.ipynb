## 1. Environment Verification

Let's verify that all required packages and tools are installed correctly.

````python
import sys
import subprocess

print("=" * 60)
print("Environment Verification")
print("=" * 60)
print()

# Python version
print(f"Python Version: {sys.version}")
print()

# Check required packages
packages_to_check = ['boto3', 'sagemaker', 'helm', 'pandas', 'numpy']

print("Installed Packages:")
for package in packages_to_check:
    try:
        module = __import__(package)
        version = getattr(module, '__version__', 'unknown')
        print(f"  ✓ {package}: {version}")
    except ImportError:
        print(f"  ✗ {package}: NOT INSTALLED")

print()

# Check gcloud
try:
    result = subprocess.run(['gcloud', '--version'], capture_output=True, text=True)
    print(f"gcloud: ✓ Installed")
    print(f"  {result.stdout.split(chr(10))[0]}")
except FileNotFoundError:
    print("gcloud: ✗ NOT FOUND")

print()

# Check uv
try:
    result = subprocess.run(['uv', '--version'], capture_output=True, text=True)
    print(f"uv: ✓ Installed")
    print(f"  {result.stdout.strip()}")
except FileNotFoundError:
    print("uv: ✗ NOT FOUND")

print()
print("=" * 60)
````

## 2. Understanding MedHELM

**MedHELM** (Medical Holistic Evaluation of Language Models) is a comprehensive benchmark for evaluating large language models in medical applications.

### Key Features:

- **Clinician-Validated Taxonomy**: 5 categories, 22 subcategories, 121 tasks
- **Realistic Medical Tasks**: From clinical decision support to patient communication
- **Safety-Focused**: Evaluates model outputs for medical safety and accuracy
- **Reproducible**: Standard scenarios and metrics for consistent evaluation

### Taxonomy Categories:

1. **Clinical Decision Support**
2. **Clinical Note Generation**
3. **Patient Communication and Education**
4. **Medical Research Assistance**
5. **Administration and Workflow**

## 3. Check Available Scenarios

Let's explore what scenarios are available in MedHELM.

````python
import os

# Check if HELM repository exists
helm_path = os.path.expanduser("~/helm")
if os.path.exists(helm_path):
    print(f"✓ HELM repository found at: {helm_path}")
    
    # Check for scenario files
    scenarios_path = os.path.join(helm_path, "src/helm/benchmark/scenarios")
    if os.path.exists(scenarios_path):
        print(f"✓ Scenarios directory found")
        print(f"\nNumber of scenario files: {len([f for f in os.listdir(scenarios_path) if f.endswith('.py')])}")
    else:
        print("✗ Scenarios directory not found")
else:
    print("✗ HELM repository not found")
    print("You may need to clone it manually: git clone https://github.com/stanford-crfm/helm.git")
````

## 4. Check Benchmark Output Directory

Verify the benchmark output directory is set up correctly.

````python
import os

benchmark_output_path = os.path.expanduser("~/benchmark_output")

if os.path.exists(benchmark_output_path):
    print(f"✓ Benchmark output directory exists: {benchmark_output_path}")
    
    # Check if any runs exist
    runs_path = os.path.join(benchmark_output_path, "runs")
    if os.path.exists(runs_path):
        runs = os.listdir(runs_path)
        print(f"\nFound {len(runs)} existing run(s):")
        for run in runs[:5]:  # Show first 5
            print(f"  - {run}")
        if len(runs) > 5:
            print(f"  ... and {len(runs) - 5} more")
    else:
        print("\nNo runs directory yet - this is normal for a fresh installation")
else:
    print(f"✗ Benchmark output directory not found: {benchmark_output_path}")
    print("Creating directory...")
    os.makedirs(benchmark_output_path, exist_ok=True)
    print("✓ Directory created")
````

## 5. AWS Configuration Check

Verify AWS credentials and SageMaker configuration.

````python
import boto3
import sagemaker

try:
    # Get SageMaker session
    session = sagemaker.Session()
    region = session.boto_region_name
    role = sagemaker.get_execution_role()
    
    print("AWS Configuration:")
    print(f"  Region: {region}")
    print(f"  Execution Role: {role}")
    print(f"  Default S3 Bucket: {session.default_bucket()}")
    print("\n✓ AWS configuration is valid")
    
except Exception as e:
    print(f"✗ Error checking AWS configuration: {str(e)}")
````
